***Motivation & Problem***

For my final project, I want to look at how people in Seattle use physical and digital library materials. As a student, I use digital resources a lot for school, but physical books are still common in libraries. This made me curious about whether people are gradually moving away from physical materials.

My main question is: How have physical and digital checkouts at the Seattle Public Library changed between 2022 and 2025?

I want to compare their monthly checkout numbers and see whether digital borrowing is growing faster than physical borrowing. I think this could help me better understand how people's reading habits are changing.

***Data***

I plan to use the Seattle Public Library's Checkouts by Title dataset from the City of Seattle Open Data Portal.

Dataset: https://data.seattle.gov/Community-and-Culture/Checkouts-by-Title/tmmm-ytt6

The dataset includes checkout counts, dates, titles, material types, and whether an item is physical or digital. This fits my question because I can compare monthly checkout numbers without collecting my own data.

The data is publicly available under the City of Seattle's Open Data Policy. Since the records do not identify individual borrowers, the privacy risk is relatively low. However, I should be careful not to assume that checkout numbers represent everyone's reading preferences, since not everyone uses public libraries.

***Unknowns & Dependencies***

One challenge might be the size of the dataset. I will probably need to filter it to only include the years I want to study. There could also be missing records or changes in library services that affect the numbers.

I also need to make sure I'm comparing similar types of materials. For example, the digital category includes more than just e-books. If the project becomes too complicated, I can narrow it down to physical books and e-books instead.
