***Overview & Research Questions***

For this project, I want to see how physical and digital borrowing at the Seattle Public Library changed from 2022 to 2025. I originally thought digital borrowing would keep increasing because people can access e-books without going to a library, but I'm not sure if that's actually happening every year.

My main research questions are:

1. How did physical and digital checkout numbers change between 2022 and 2025?

2. Did digital borrowing grow faster than physical borrowing during these four years?

3. Was there a noticeable change in checkout numbers during the library's system outage in 2024?

I expect digital checkouts to increase overall, but I also want to see if the growth has started to slow down. I'm especially interested in whether physical books are still being borrowed frequently, even with more digital options available.

***Background & Related Work***

I found some useful information in the Seattle Public Library's annual reports.

According to the library's 2024 Annual Levy Report, total checkouts dropped from 13.4 million in 2023 to 10.8 million in 2024. One important reason was the ransomware attack in May 2024, which disrupted library services. E-books were unavailable for almost three weeks, and other services took longer to recover.

This is important for my project because if I see a sudden decrease around mid-2024, it might not mean people stopped borrowing books. They might simply have been unable to access the library's services.

The library's 2025 Impact Report also shows that digital materials were checked out 7.2 million times, compared with 5.5 million physical checkouts. This suggests digital borrowing is an important part of library use, but I still want to see how the monthly numbers changed.

These reports give me some background, but my project will focus more on monthly trends instead of only looking at yearly totals.

***Methodology***

I plan to use the Seattle Public Library's Checkouts by Title dataset for my analysis. Since the dataset is very large, I will follow my professor's suggestion and use the Socrata API instead of downloading the entire dataset.

I will only look at data from 2022 to 2025. I plan to group the records by year, month, UsageClass, and MaterialType, and add up the checkout counts for each group.

First, I will check the available material types and make sure the data is complete for the years I want to study. I also want to see if I can compare physical books and e-books separately, since comparing all physical and digital materials might include things like music and movies.

After preparing the data, I plan to use Python and pandas to calculate monthly and yearly checkout totals. I will make line graphs to show how borrowing changes over time and bar charts to compare physical and digital checkouts. I will also calculate percentage changes to see which category grew faster.

I will pay special attention to May through September 2024 because of the ransomware attack. If the numbers drop during that period, I will compare them with other months and explain how the service disruption might have affected the results.

One limitation is that checkout numbers don't tell us exactly why people choose digital or physical materials. I can describe borrowing trends, but I can't directly measure people's reading preferences using this dataset alone.
